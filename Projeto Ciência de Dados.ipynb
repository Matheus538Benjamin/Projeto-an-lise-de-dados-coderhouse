## 6. Análise Exploratória de Dados (EDA)

Nesta etapa, buscamos entender melhor os dados por meio de visualizações gráficas e estatísticas descritivas. O objetivo é identificar padrões, correlações, comportamentos relevantes e possíveis outliers que ajudem na interpretação e modelagem dos dados.



### 6.1 Evolução do Consumo Global de Energia

O primeiro gráfico apresenta a evolução do consumo de energia ao longo dos anos. Essa visualização permite observar tendências globais no uso de energia, como crescimento contínuo, estabilidade ou possíveis quedas em determinados períodos. É uma análise temporal que ajuda a contextualizar o comportamento geral dos dados.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Agrupar por ano e somar o consumo total
df_ano = df_energia.groupby("Ano")["Consumo total de energia (TWH)"].sum().reset_index()

# Plotar
plt.figure(figsize=(12,6))
sns.lineplot(data=df_ano, x="Ano", y="Consumo total de energia (TWH)", marker="o")
plt.title("Evolução do Consumo Global de Energia ao Longo dos Anos")
plt.xlabel("Ano")
plt.ylabel("Consumo Total de Energia (TWH)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

### 6.2 Top 10 Países com Maior Uso de Energia Renovável

Aqui, identificamos os 10 países com maior percentual de uso de energia renovável no último ano disponível no dataset. Este gráfico nos ajuda a entender quais países estão mais avançados na transição energética e comprometidos com práticas sustentáveis.


```python
# Último ano disponível
ano_recente = df_energia["Ano"].max()

# Top 10 países com maior participação de energia renovável
top10_renov = df_energia[df_energia["Ano"] == ano_recente] \
    .sort_values("Compartilhamento de energia renovável (%)", ascending=False).head(10)

# Verificar se a seleção deu certo
print(top10_renov[["País", "Compartilhamento de energia renovável (%)"]])

# Plotar
plt.figure(figsize=(12,6))
sns.barplot(data=top10_renov, x="Compartilhamento de energia renovável (%)", y="País", palette="viridis")
plt.title(f"Top 10 Países com Maior Uso de Energia Renovável ({ano_recente})")
plt.xlabel("Energia Renovável (%)")
plt.ylabel("País")
plt.tight_layout()
plt.show()

```

### 6.3 Matriz de Correlação Geral

Nesta visualização, exploramos as correlações entre variáveis numéricas relevantes do conjunto de dados completo, considerando todos os anos disponíveis. Essa matriz de correlação geral nos permite observar relações lineares médias entre o consumo de energia per capita, as emissões de carbono, o índice de preços de energia e a participação de fontes renováveis. O heatmap facilita a identificação de padrões de associação entre as variáveis ao longo do tempo.



```python


# Seleciona as colunas numéricas relevantes
colunas_interesse = [
    "Uso de energia per capita (kWh)",
    "Emissões de carbono (milhões de toneladas)",
    "Índice de Preços de Energia (USD/KWH)",
    "Compartilhamento de energia renovável (%)"
]

# Filtra o DataFrame com essas colunas
df_corr = df_energia[colunas_interesse]

# Calcula a matriz de correlação
matriz_corr = df_corr.corr()

# Plota o heatmap
plt.figure(figsize=(8,6))
sns.heatmap(matriz_corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Matriz de Correlação Geral (Todos os Anos)")
plt.tight_layout()
plt.show()


```


### 6.4 Distribuição do Índice de Preços de Energia

O último gráfico desta seção é um histograma que mostra como o índice de preços de energia (USD/KWH) está distribuído entre os países. ### 6.5 Análise de Outliers

Para complementar a análise exploratória, é importante investigar a presença de outliers nas variáveis numéricas. Outliers podem distorcer médias, influenciar modelos estatísticos e indicar dados inconsistentes ou casos excepcionais.

Nesta etapa, utilizamos boxplots para visualizar a distribuição das variáveis e identificar possíveis valores fora do padrão. Essa verificação é útil para decidir se será necessário aplicar tratamentos adicionais, como normalização, transformação logarítmica ou remoção de valores extremos.

### 7. Transformação e Preparação dos Dados

Nesta etapa, foi realizada uma verificação básica da qualidade da base de dados, focando principalmente na presença de valores nulos e nos tipos de dados de cada coluna.

Para garantir consistência nas análises, os valores nulos encontrados em colunas numéricas foram preenchidos com a média da respectiva coluna.

Essas ações são fundamentais para evitar erros nas próximas etapas e garantir que os gráficos e análises não sofram distorções.



```python
# Verificar se tem valores nulos
print("Valores nulos por coluna:")
print(df_energia.isnull().sum())

# Verificar os tipos de dados
print("\nTipos de dados por coluna:")
print(df_energia.dtypes)

# Se tiver nulos em colunas numéricas, preenche com a média
colunas_numericas = df_energia.select_dtypes(include=["float64", "int64"]).columns

for coluna in colunas_numericas:
    if df_energia[coluna].isnull().sum() > 0:
        media = df_energia[coluna].mean()
        df_energia[coluna].fillna(media, inplace=True)
        print(f"Valores nulos preenchidos com a média na coluna: {coluna}")
```

### 8. Aplicação de PCA (Análise de Componentes Principais)

A Análise de Componentes Principais (PCA) foi utilizada para reduzir a dimensionalidade do conjunto de dados, facilitando a visualização de padrões e a identificação de possíveis agrupamentos.

Os dados foram previamente padronizados com `StandardScaler`, garantindo que todas as variáveis tivessem a mesma importância na análise.

A projeção em dois componentes principais permitiu visualizar como os países se distribuem no espaço reduzido, com base nas variáveis energéticas selecionadas.


```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# 1. Selecionar as colunas numéricas
colunas_numericas = df_energia.select_dtypes(include=["float64", "int64"]).columns

# 2. Escalonar os dados (PCA precisa disso!)
scaler = StandardScaler()
dados_escalonados = scaler.fit_transform(df_energia[colunas_numericas])

# 3. Aplicar PCA
pca = PCA(n_components=2)  # vamos reduzir para 2 dimensões
componentes = pca.fit_transform(dados_escalonados)

# 4. Criar um DataFrame com os componentes principais
df_pca = pd.DataFrame(componentes, columns=["PC1", "PC2"])
df_pca["País"] = df_energia["País"].values
df_pca["Ano"] = df_energia["Ano"].values

# 5. Visualizar
plt.figure(figsize=(10,6))
sns.scatterplot(data=df_pca, x="PC1", y="PC2", hue="Ano", alpha=0.7, palette="viridis")
plt.title("PCA - Projeção dos Dados em 2 Componentes Principais")
plt.xlabel("Componente Principal 1")
plt.ylabel("Componente Principal 2")
plt.tight_layout()
plt.show()
```

### 9.2 Diferenças Regionais no Uso de Energia

Países desenvolvidos tendem a ter um uso de energia per capita significativamente maior. Já os países em desenvolvimento apresentam valores menores, mas com crescimento acelerado.

```python
# Pega o último ano de cada país
df_ultimo_ano = df_energia.sort_values('Ano').groupby('País').tail(1)

plt.figure(figsize=(12,6))
sns.barplot(data=df_ultimo_ano.sort_values('Uso de energia per capita (kWh)', ascending=False),
            x='País', y='Uso de energia per capita (kWh)', palette='viridis')
plt.title('Uso de Energia Per Capita por País (Último Ano)')
plt.xlabel('País')
plt.ylabel('Uso de energia per capita (kWh)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

O gráfico evidencia uma **grande disparidade no uso de energia per capita** entre países. Nações desenvolvidas consomem muito mais energia por habitante, reflexo de estilos de vida com alto consumo e infraestrutura mais dependente de energia. Já os países em desenvolvimento, embora apresentem consumo per capita menor, mostram **tendência de crescimento**, o que representa um **desafio futuro** em termos de sustentabilidade energética.

### 9.3 Adoção de Energias Renováveis

A porcentagem de energia renovável ainda é pequena na maioria dos países, embora alguns estejam avançando rapidamente nesse quesito. A análise mostra que a transição energética é desigual, refletindo políticas públicas e capacidades tecnológicas distintas.

```python
plt.figure(figsize=(12,6))
sns.barplot(data=df_ultimo_ano.sort_values('Compartilhamento de energia renovável (%)', ascending=False),
            x='País', y='Compartilhamento de energia renovável (%)', palette='crest')
plt.title('Compartilhamento de Energia Renovável por País (Último Ano)')
plt.xlabel('País')
plt.ylabel('Energia Renovável (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Apesar de avanços pontuais, a **adoção de energias renováveis** ainda é baixa na maioria dos países. O gráfico mostra que poucos países lideram essa transição, indicando uma **distribuição desigual** de políticas, investimentos e acesso a tecnologias verdes. Isso revela a importância de **iniciativas globais de cooperação** para acelerar a transição energética e reduzir a dependência de fontes poluentes.

### 9.4 Gráfico de Correlação entre Dependência de Combustíveis Fósseis e Emissões de Carbono

O gráfico gerado foi um **heatmap** que exibe a **correlação** entre as variáveis **Dependência de Combustíveis Fósseis (%)** e **Emissões de Carbono (milhões de toneladas)** para **cada país**. A análise é baseada na **soma total dos valores** de cada variável ao longo de todos os anos disponíveis, ou seja, para cada país, foram somados os valores de **Dependência de Combustíveis Fósseis** e **Emissões de Carbono** ao longo dos anos (de 2010 a 2024).


```python
# Somar os valores de 'Dependência de combustível fóssil' e 'Emissões de carbono' por país
df_energia_somada = df_energia.groupby('País')[['Dependência de combustível fóssil (%)', 'Emissões de carbono (milhões de toneladas)']].sum()

# Plotando o heatmap com a soma dos anos para cada país
plt.figure(figsize=(14,8))
sns.heatmap(df_energia_somada.corr(), annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)

# Ajustes do gráfico
plt.title('Correlação entre Dependência de Combustíveis Fósseis e Emissões de Carbono (Somado por País)', fontsize=16)
plt.tight_layout()
plt.show()
```

### Análise da Correlação entre Dependência de Combustíveis Fósseis e Emissões de Carbono

Na análise realizada, foi calculada a **correlação entre a dependência de combustíveis fósseis** e as **emissões de carbono** para cada país, considerando a **soma dos valores** ao longo dos anos. A correlação obtida foi de **0.92**, o que indica uma **forte correlação positiva** entre essas duas variáveis.

#### O que isso significa?

Uma correlação de **0.92** sugere que:

- Existe uma **relação direta e forte** entre a **dependência de combustíveis fósseis** e as **emissões de carbono**. Ou seja, países com uma maior dependência de combustíveis fósseis tendem a gerar **mais emissões de carbono**.
- Este padrão é esperado, pois os **combustíveis fósseis**, como o carvão, petróleo e gás natural, são as principais fontes de **emissões de gases de efeito estufa**.
  
  
### Conclusão

A correlação positiva de **0.92** entre a dependência de combustíveis fósseis e as emissões de carbono evidencia a relação crítica entre essas duas variáveis e destaca a necessidade de estratégias mais eficazes para a **descarbonização** da matriz energética global.
