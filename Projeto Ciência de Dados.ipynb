## 8. Aplicação de PCA (Análise de Componentes Principais)

A Análise de Componentes Principais (PCA) foi utilizada para reduzir a dimensionalidade do conjunto de dados, facilitando a visualização de padrões e a identificação de possíveis agrupamentos.

Os dados foram previamente padronizados com `StandardScaler`, garantindo que todas as variáveis tivessem a mesma importância na análise.

A projeção em dois componentes principais permitiu visualizar como os países se distribuem no espaço reduzido, com base nas variáveis energéticas selecionadas.


```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# 1. Selecionar as colunas numéricas
colunas_numericas = df_energia.select_dtypes(include=["float64", "int64"]).columns

# 2. Escalonar os dados (PCA precisa disso!)
scaler = StandardScaler()
dados_escalonados = scaler.fit_transform(df_energia[colunas_numericas])

# 3. Aplicar PCA
pca = PCA(n_components=2)  # vamos reduzir para 2 dimensões
componentes = pca.fit_transform(dados_escalonados)

# 4. Criar um DataFrame com os componentes principais
df_pca = pd.DataFrame(componentes, columns=["PC1", "PC2"])
df_pca["País"] = df_energia["País"].values
df_pca["Ano"] = df_energia["Ano"].values

# 5. Visualizar
plt.figure(figsize=(10,6))
sns.scatterplot(data=df_pca, x="PC1", y="PC2", hue="Ano", alpha=0.7, palette="viridis")
plt.title("PCA - Projeção dos Dados em 2 Componentes Principais")
plt.xlabel("Componente Principal 1")
plt.ylabel("Componente Principal 2")
plt.tight_layout()
plt.show()
```

### 9.6 – Análise de Correlação Individual com Regressão Linear Simples

Nesta etapa exploratória, foi realizada uma análise visual para verificar a **força da relação entre o consumo total de energia** e três variáveis que **não estão expressas em percentual**:

- Uso de energia per capita (kWh)
- Emissões de carbono (milhões de toneladas)
- Índice de Preços de Energia (USD/KWH)

#### 📈 O que foi feito:

Para cada variável explicativa:
1. Foi gerado um gráfico de dispersão comparando os dados reais com a linha de uma regressão linear simples.
2. O coeficiente de determinação (**R²**) foi calculado para indicar o nível de explicação da variável sobre o consumo energético.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Target
target = 'Consumo total de energia (TWH)'

# Variáveis específicas que você quer analisar
variaveis_explicativas = [
    'Uso de energia per capita (kWh)',
    'Emissões de carbono (milhões de toneladas)',
    'Índice de Preços de Energia (USD/KWH)'
]

# Imprime as colunas para conferência
print("Colunas no df_ano:", df_ano.columns.tolist())

# Filtra só as colunas que existem mesmo no DataFrame para evitar erro
variaveis_existentes = [var for var in variaveis_explicativas if var in df_ano.columns]

if not variaveis_existentes:
    print("Nenhuma das variáveis especificadas está presente no DataFrame.")
else:
    for var in variaveis_existentes:
        X = df_ano[[var]].values
        y = df_ano[target].values
        
        modelo = LinearRegression()
        modelo.fit(X, y)
        y_pred = modelo.predict(X)
        r2 = r2_score(y, y_pred)
        
        plt.figure(figsize=(8, 5))
        sns.scatterplot(x=X.flatten(), y=y, color='blue', label='Dados reais')
        plt.plot(X, y_pred, color='red', label=f'Regressão linear (R² = {r2:.2f})')
        plt.xlabel(var)
        plt.ylabel(target)
        plt.title(f'{var} vs {target}')
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.show()


```

#### 🎯 Conclusão desta etapa:

Como todas as três variáveis apresentaram **algum grau de correlação** com o consumo total, optamos por utilizá-las em conjunto em uma **regressão linear múltipla**, consolidando as três como variáveis independentes para prever a variável target (`Consumo total de energia`).

## 10. Treinar um modelo de regressão linear com todas as variáveis

Neste passo, faremos um teste inicial utilizando **regressão linear múltipla** com 3 variáveis independentes para prever o consumo total de energia por ano.

🎯 Nosso objetivo aqui é entender se essas variáveis explicam bem o consumo total, como um **modelo de regressão base**.

As variáveis escolhidas foram:

- Uso de energia per capita (kWh)
- Emissões de carbono (milhões de toneladas)
- Índice de Preços de Energia (USD/KWH)

A variável alvo (target) será:

- Consumo total de energia (TWH)


### 10.1 Agrupar os dados por ano

Vamos somar os dados numéricos por ano para gerar uma base consolidada anual.


```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Agrupar os dados por ano (somando os valores numéricos)
df_ano = df_energia.groupby('Ano').sum(numeric_only=True).reset_index()
```

### 10.2 – Definir as variáveis de entrada (X) e saída (y)

Selecionamos as 3 variáveis que servirão como preditoras, e o consumo total será a variável alvo.


```python
# Variável alvo
target = 'Consumo total de energia (TWH)'

# Variáveis explicativas
features = [
    'Uso de energia per capita (kWh)',
    'Emissões de carbono (milhões de toneladas)',
    'Índice de Preços de Energia (USD/KWH)'
]

# Separar X e y
X = df_ano[features]
y = df_ano[target]
```

### 10.3 – Criar e treinar o modelo de Regressão Linear

Nesta etapa, criamos o modelo e treinamos com todos os dados disponíveis.


```python
# Criar e treinar o modelo de regressão linear múltipla
modelo = LinearRegression()
modelo.fit(X, y)
```

### 10.4 – Fazer previsões e calcular métricas

Vamos avaliar o quão bem o modelo consegue prever os valores de y com base nas variáveis X.


```python
# Fazer a predição com os próprios dados
y_pred = modelo.predict(X)

# Avaliar o desempenho
r2 = r2_score(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))
```

### 10.5 Mostrar os coeficientes da equação de regressão

Isso nos permite ver qual o impacto de cada variável na previsão do consumo.


```python
# Mostrar os coeficientes da equação
print("🔢 Coeficientes:")
for f, c in zip(features, modelo.coef_):
    print(f"- {f}: {c:.2f}")

print(f"\n📈 Intercepto: {modelo.intercept_:.2f}")
print(f"📊 R² da regressão múltipla: {r2:.4f}")
print(f"📉 RMSE: {rmse:.2f}")
```

### 10.6 Visualização gráfica com a equação e o R²

Aqui comparamos os valores reais e previstos ao longo dos anos.


```python
# Plotando os valores reais vs previstos com R² e coeficientes no título
plt.figure(figsize=(10, 6))
sns.lineplot(x=df_ano['Ano'], y=y, label='Real', marker='o')
sns.lineplot(x=df_ano['Ano'], y=y_pred, label='Previsto', marker='s')

# Montar equação do modelo no título
coef_text = " + ".join([f"{c:.2f}×{f}" for f, c in zip(features, modelo.coef_)])
plt.title(f"Regressão Linear Múltipla: {target}\n{coef_text} + {modelo.intercept_:.2f} | R² = {r2:.2f}")

plt.xlabel("Ano")
plt.ylabel("Consumo de Energia (TWH)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

conseguem prever o **Consumo total de energia (TWH)** ao longo dos anos.

- O R² (qualidade da explicação)
- O RMSE (erro médio quadrático)

Este é um **primeiro experimento exploratório**, que serve como base para os próximos testes com algoritmos mais robustos.



### 10.8 Comparação entre Modelos de Regressão (com dados de teste)

Neste bloco, comparamos o desempenho de três algoritmos de regressão:

- Regressão Linear
- Random Forest Regressor
- XGBoost Regressor

Todos os modelos foram treinados previamente com os dados de treino (`X_train`, `y_train`)  
e aqui vamos:

- Prever os valores de `y_test`
- Calcular R² e RMSE de cada modelo
- Escolher automaticamente o modelo com melhor R²


```python
# 📦 Importações
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import r2_score, mean_squared_error

# 🧮 Agrupar por ano e somar
df_ano = df_energia.groupby('Ano').sum(numeric_only=True).reset_index()

# 🎯 Selecionar features e target
X_full = df_ano[[ 
    'Uso de energia per capita (kWh)', 
    'Emissões de carbono (milhões de toneladas)', 
    'Índice de Preços de Energia (USD/KWH)' 
]]
y_full = df_ano['Consumo total de energia (TWH)']

# 🔀 Separar dados em treino e teste
X_train, X_test, y_train, y_test = train_test_split(
    X_full, y_full, test_size=0.3, random_state=42
)

# 🤖 Treinar os modelos
modelo_lr = LinearRegression().fit(X_train, y_train)
modelo_rf = RandomForestRegressor(random_state=42).fit(X_train, y_train)
modelo_xgb = XGBRegressor(random_state=42).fit(X_train, y_train)

# 📚 Dicionário com os modelos treinados
modelos = {
    'Linear Regression': modelo_lr,
    'Random Forest': modelo_rf,
    'XGBoost': modelo_xgb
}

# 📊 Avaliar desempenho (R² e RMSE)
resultados = {}
for nome, mod in modelos.items():
    y_pred = mod.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    
    resultados[nome] = {'R²': round(r2, 4), 'RMSE': round(rmse, 2)}

# 🖨️ Exibir resultados
print("📊 Resultados de R² e RMSE:")
for nome, res in resultados.items():
    print(f"{nome}: R² = {res['R²']} | RMSE = {res['RMSE']}")

# 🥇 Escolher o melhor modelo com maior R²
melhor_modelo_nome = max(resultados, key=lambda x: resultados[x]['R²'])
melhor_modelo = modelos[melhor_modelo_nome]

print(f"\n✅ Melhor modelo selecionado: {melhor_modelo_nome}")
```

Neste bloco, usamos **regressão linear múltipla** com 3 variáveis fixas para prever o consumo total de energia (TWH). Porém, diferentemente dos testes anteriores, aplicamos aqui uma abordagem de **validação repetida com divisão aleatória dos dados**.

O objetivo é:
- Rodar **100 divisões diferentes (train/test aleatórios)**
- Avaliar a performance de cada split
- Escolher o modelo que alcançar o **melhor R²**

Esse processo simula uma espécie de **"busca estocástica" pelo melhor cenário de treinamento**, usando apenas regressão linear.


### 11.1 Preparar os dados

Agrupamos os dados por ano e separamos as variáveis explicativas e a variável alvo.


```python
# Agrupar por ano e somar
df_ano = df_energia.groupby('Ano').sum(numeric_only=True).reset_index()

# Features e target
X_full = df_ano[[
    'Uso de energia per capita (kWh)',
    'Emissões de carbono (milhões de toneladas)',
    'Índice de Preços de Energia (USD/KWH)'
]]

y_full = df_ano['Consumo total de energia (TWH)']
```

### 11.2 Rodar múltiplos testes aleatórios

Vamos realizar 100 divisões aleatórias entre treino e teste, e para cada uma:
- Treinar um modelo de regressão linear
- Calcular R² e RMSE
- Armazenar o resultado
- Guardar o melhor modelo (com maior R²)


```python
# Variáveis para armazenar resultados
resultados = []
melhor_r2 = -np.inf
melhor_modelo = None
melhor_split = {}

# Rodar 100 divisões aleatórias
for i in range(100):
    X_train, X_test, y_train, y_test = train_test_split(
        X_full, y_full, test_size=0.3, random_state=np.random.randint(1000)
    )

    modelo = LinearRegression()
    modelo.fit(X_train, y_train)
    y_pred = modelo.predict(X_test)

    r2 = r2_score(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))

    resultados.append({'R²': r2, 'RMSE': rmse})

    if r2 > melhor_r2:
        melhor_r2 = r2
        melhor_modelo = modelo
        melhor_split = {
            'X_test': X_test,
            'y_test': y_test,
            'y_pred': y_pred
        }
```

### 13.1 – Preparação dos Dados Reais para Modelagem

Nesta etapa, iniciamos a construção da previsão de consumo energético utilizando a base de dados real (`df_energia`). O objetivo é preparar os dados para alimentar os modelos de regressão que serão comparados mais adiante.

#### 📌 O que foi feito:

2. **Definição das Variáveis para o Modelo**

     - Uso de energia per capita (kWh)
     - Emissões de carbono (milhões de toneladas)
     - Índice de Preços de Energia (USD/KWH)
   - `y_full`: variável dependente (target), que representa o **Consumo total de energia (TWH)** por ano.

```python
# Agrupar por ano se necessário
df_ano = df_energia.groupby('Ano').sum(numeric_only=True).reset_index()

# Definir X e y com suas variáveis reais
X_full = df_ano[[
    'Uso de energia per capita (kWh)',
    'Emissões de carbono (milhões de toneladas)',
    'Índice de Preços de Energia (USD/KWH)'
]]
y_full = df_ano['Consumo total de energia (TWH)']
```

### 13.2 – Função para Encontrar o Melhor Split

Criamos uma função chamada `encontrar_melhor_split` que executa o seguinte:

- Realiza **100 divisões aleatórias** dos dados (`train_test_split`)
- Treina o modelo especificado
- Calcula o **R²** e o **RMSE** em cada split
- Guarda o modelo com o **melhor R²**

Essa função serve para escolher o modelo que teve o melhor desempenho entre as 100 tentativas.


```python
def encontrar_melhor_split(X_full, y_full, modelo_classe, nome_modelo, n_testes=100):
    melhor_r2 = -np.inf
    melhor_modelo = None
    melhor_split = {}

    for _ in range(n_testes):
        X_train, X_test, y_train, y_test = train_test_split(
            X_full, y_full, test_size=0.3, random_state=np.random.randint(1000)
        )

        modelo = modelo_classe()
        modelo.fit(X_train, y_train)
        y_pred = modelo.predict(X_test)

        r2 = r2_score(y_test, y_pred)
        rmse = np.sqrt(mean_squared_error(y_test, y_pred))

        if r2 > melhor_r2:
            melhor_r2 = r2
            melhor_modelo = modelo
            melhor_split = {
                'nome': nome_modelo,
                'X_test': X_test,
                'y_test': y_test,
                'y_pred': y_pred,
                'r2': r2,
                'rmse': rmse,
                'modelo': modelo
            }

    return melhor_split
```

### 13.3 – Seleção do Melhor Modelo por Tipo

Aplicamos a função `encontrar_melhor_split` para cada um dos três modelos:

- Regressão Linear
- Random Forest
- XGBoost

O objetivo aqui foi obter o **melhor modelo (top 1)** de cada tipo, com base no maior valor de R² entre 100 divisões aleatórias dos dados.


```python
# Buscar o top 1 de cada modelo
melhor_lr = encontrar_melhor_split(X_full, y_full, LinearRegression, 'Linear Regression')
melhor_rf = encontrar_melhor_split(X_full, y_full, RandomForestRegressor, 'Random Forest')
melhor_xgb = encontrar_melhor_split(X_full, y_full, XGBRegressor, 'XGBoost')
```

### 13.4 – Exibição das Métricas dos Melhores Modelos

Após encontrar o melhor split de cada modelo, mostramos na tela os seguintes indicadores:

- **R²**: quanto o modelo consegue explicar da variação do consumo
- **RMSE**: erro médio das previsões em relação aos valores reais

Com isso, conseguimos comparar rapidamente o desempenho dos três modelos.


```python
print("🏆 Melhores Modelos Selecionados (Top 1 de cada tipo):")
for modelo in [melhor_lr, melhor_rf, melhor_xgb]:
    print(f"{modelo['nome']} → R²: {modelo['r2']:.4f} | RMSE: {modelo['rmse']:.2f}")
```

Nesta etapa, os três modelos foram treinados com **100% da base de dados**.

```python
modelo_lr = LinearRegression().fit(X_full, y_full)
modelo_rf = RandomForestRegressor().fit(X_full, y_full)
modelo_xgb = XGBRegressor().fit(X_full, y_full)

# Previsões com os 3 modelos
y_pred_lr = modelo_lr.predict(X_full)
y_pred_rf = modelo_rf.predict(X_full)
y_pred_xgb = modelo_xgb.predict(X_full)

# Criar tabela comparativa
tabela_comparativa = pd.DataFrame({
    'Ano': df_ano['Ano'],
    'Consumo Real (TWH)': y_full,
    'Linear Regression (TWH)': y_pred_lr,
    'Random Forest (TWH)': y_pred_rf,
    'XGBoost (TWH)': y_pred_xgb
})

# Calcular erro percentual
tabela_comparativa['Erro LR (%)'] = (
    (tabela_comparativa['Linear Regression (TWH)'] - tabela_comparativa['Consumo Real (TWH)']) /
    tabela_comparativa['Consumo Real (TWH)']) * 100

tabela_comparativa['Erro RF (%)'] = (
    (tabela_comparativa['Random Forest (TWH)'] - tabela_comparativa['Consumo Real (TWH)']) /
    tabela_comparativa['Consumo Real (TWH)']) * 100

tabela_comparativa['Erro XGB (%)'] = (
    (tabela_comparativa['XGBoost (TWH)'] - tabela_comparativa['Consumo Real (TWH)']) /
    tabela_comparativa['Consumo Real (TWH)']) * 100

# Arredondar resultados
tabela_comparativa = tabela_comparativa.round(2)

# Exibir a tabela
print("📋 Comparativo de Previsões e Erros Percentuais:")
display(tabela_comparativa)
```

```python
# R² e RMSE para cada modelo na base completa
r2_lr = r2_score(y_full, tabela_comparativa['Linear Regression (TWH)'])
r2_rf = r2_score(y_full, tabela_comparativa['Random Forest (TWH)'])
r2_xgb = r2_score(y_full, tabela_comparativa['XGBoost (TWH)'])

rmse_lr = mean_squared_error(y_full, tabela_comparativa['Linear Regression (TWH)'], squared=False)
rmse_rf = mean_squared_error(y_full, tabela_comparativa['Random Forest (TWH)'], squared=False)
rmse_xgb = mean_squared_error(y_full, tabela_comparativa['XGBoost (TWH)'], squared=False)

# Erro percentual médio absoluto
erro_medio_lr = tabela_comparativa['Erro LR (%)'].abs().mean()
erro_medio_rf = tabela_comparativa['Erro RF (%)'].abs().mean()
erro_medio_xgb = tabela_comparativa['Erro XGB (%)'].abs().mean()

# Montar DataFrame comparativo
comparativo_modelos = pd.DataFrame({
    'Modelo': ['Linear Regression', 'Random Forest', 'XGBoost'],
    'Erro Médio (%)': [erro_medio_lr, erro_medio_rf, erro_medio_xgb],
    'R²': [r2_lr, r2_rf, r2_xgb],
    'RMSE': [rmse_lr, rmse_rf, rmse_xgb]
}).round(3)

# Determinar o melhor modelo com base no menor erro percentual
melhor_modelo_nome = comparativo_modelos.loc[comparativo_modelos['Erro Médio (%)'].idxmin(), 'Modelo']

# Exibir tabela e resultado
from IPython.display import display
display(comparativo_modelos)

print(f"\n🏆 Melhor modelo com base no menor erro percentual médio: **{melhor_modelo_nome}**")
```

# 14. Validação de Modelos com Novas Estratégias

Após observarmos que o modelo **XGBoost apresentou desempenho perfeito** (R² = 1.0, erro médio = 0%, RMSE ≈ 1.0) ao ser treinado e avaliado sobre **100% da base de dados**, decidimos seguir com uma nova etapa para tornar nossa análise mais **realista e confiável**.

📌 **Objetivo desta nova fase:**  
Aplicar **métodos de validação e divisão dos dados** para investigar se os modelos continuam apresentando bons resultados em contextos de **generalização** – ou seja, em dados **não vistos durante o treinamento**.


#### 14.1 Primeira Avaliação com Ridge e Lasso (Divisão 70/30)

Nesta etapa inicial, aplicamos duas técnicas de regressão regularizada — **Ridge** (penalização L2) e **Lasso** (penalização L1) — com hiperparâmetros definidos manualmente.

📌 **O que foi feito:**
- A base de dados foi dividida em dois conjuntos:
  - **70% para treino**
  - **30% para teste**
- Foram utilizados os seguintes modelos:
  - `Ridge(alpha=1.0)`: aplica penalização L2 aos coeficientes, suavizando os pesos para evitar overfitting.
  - `Lasso(alpha=0.1)`: aplica penalização L1, que além de suavizar, pode reduzir alguns coeficientes a zero, promovendo seleção de variáveis.


```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge, Lasso

# Divisão dos dados em treino (70%) e teste (30%)
X_train, X_test, y_train, y_test = train_test_split(
    X_full, y_full, test_size=0.3, random_state=42
)

# Ridge Regression (penalização L2)
ridge = Ridge(alpha=1.0)
ridge.fit(X_train, y_train)
r2_ridge = ridge.score(X_test, y_test)
print(f"R² Ridge: {r2_ridge:.4f}")

# Lasso Regression (penalização L1)
lasso = Lasso(alpha=0.1)
lasso.fit(X_train, y_train)
r2_lasso = lasso.score(X_test, y_test)
print(f"R² Lasso: {r2_lasso:.4f}")
```

##### 14.1.1 Comparação do R² em Diferentes Proporções de Teste

Para entender como os modelos **Ridge** e **Lasso** se comportam com diferentes proporções de dados reservados para teste, realizamos uma análise variando o parâmetro `test_size` de 20% até 40%.

📊 **Resultados Obtidos:**

| Test Size | R² Ridge | R² Lasso |
|-----------|----------|----------|
| 0.20      | 0.4465   | 0.4454   |
| 0.25      | 0.3104   | 0.3088   |
| 0.30      | 0.0765   | 0.0757   |
| 0.35      | 0.1006   | 0.1000   |
| 0.40      | 0.0054   | 0.0035   |

📌 **O que significa esse resultado?**

- O **R²** indica o quanto o modelo consegue explicar a variabilidade dos dados de teste. Quanto mais próximo de **1.0**, melhor.
- À medida que **aumentamos a porcentagem de dados reservados para teste**, o desempenho dos modelos **cai significativamente**.
- Isso mostra que os modelos **não estão generalizando bem**: eles funcionam melhor quando têm **mais dados para aprender (treino)** e pioram conforme têm **menos dados para treino e mais dados para prever (teste)**.

🔎 **Conclusão prática:**
- Tanto o **Ridge** quanto o **Lasso** apresentam **baixa capacidade de generalização** nesse conjunto de dados sem ajuste fino.
- O desempenho mais estável foi com **20% de teste**, mas já com resultados modestos (R² ~0.44).
- Conforme o teste aumenta, os modelos **perdem poder preditivo**, chegando a quase **não explicar nada (R² ≈ 0.00)** com 40% de teste.


### 14.2 Ajuste de Hiperparâmetros com GridSearchCV (Ridge e Lasso)

Nesta etapa, utilizamos o **GridSearchCV** para ajustar o hiperparâmetro **`alpha`** dos modelos **Ridge** e **Lasso**, avaliando o desempenho em diferentes divisões de treino e teste.

📌 **Objetivo**: Encontrar o melhor valor de `alpha` (regularização) para cada modelo, usando validação cruzada (5 folds), e depois verificar o desempenho no conjunto de teste.

🔧 **Detalhes da abordagem:**
- Usamos 20 valores de `alpha`, variando de `1e-4` até `1e2`, em escala logarítmica.
- Para cada valor de `test_size` (20% a 40%), o processo é repetido:
  - Divisão dos dados (`train_test_split`)
  - Treinamento do modelo com validação cruzada (`GridSearchCV`)
  - Avaliação do melhor modelo no conjunto de teste.

📊 **Métricas observadas:**
- **Melhor alpha**: valor de regularização ideal encontrado via validação cruzada.
- **R² (validação cruzada)**: desempenho médio durante o GridSearch.
- **R² (teste)**: desempenho do modelo final em dados nunca vistos.

✅ **Vantagem dessa abordagem**:
- Garante uma busca mais robusta por modelos generalizáveis.
- Mostra como o desempenho pode variar conforme a proporção dos dados de teste e o ajuste fino dos hiperparâmetros.


```python
import numpy as np
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.linear_model import Ridge, Lasso

# Alphas: de 1e-4 até 1e2, com mais granularidade
alphas = np.logspace(-4, 2, 20)

# Test sizes para avaliar
test_sizes = [0.20, 0.25, 0.30, 0.35, 0.40]

for size in test_sizes:
    print(f"\nTest size: {int(size * 100)}%")

    # Divisão dos dados
    X_train, X_test, y_train, y_test = train_test_split(
        X_full, y_full, test_size=size, random_state=42
    )

    # === Ridge Regression ===
    ridge = Ridge(random_state=42)
    grid_ridge = GridSearchCV(
        estimator=ridge,
        param_grid={'alpha': alphas},
        cv=5,
        scoring='r2',
        n_jobs=-1
    )
    grid_ridge.fit(X_train, y_train)
    best_ridge = grid_ridge.best_estimator_
    r2_ridge_test = best_ridge.score(X_test, y_test)

    print(f"  🔷 Ridge | Melhor alpha: {grid_ridge.best_params_['alpha']}")
    print(f"     R² (validação cruzada): {grid_ridge.best_score_:.4f}")
    print(f"     R² (teste): {r2_ridge_test:.4f}")

    # === Lasso Regression ===
    lasso = Lasso(random_state=42, max_iter=10000)
    grid_lasso = GridSearchCV(
        estimator=lasso,
        param_grid={'alpha': alphas},
        cv=5,
        scoring='r2',
        n_jobs=-1
    )
    grid_lasso.fit(X_train, y_train)
    best_lasso = grid_lasso.best_estimator_
    r2_lasso_test = best_lasso.score(X_test, y_test)

    print(f"  🔶 Lasso | Melhor alpha: {grid_lasso.best_params_['alpha']}")
    print(f"     R² (validação cruzada): {grid_lasso.best_score_:.4f}")
    print(f"     R² (teste): {r2_lasso_test:.4f}")


```

##### 14.2.1 Resultados do Ajuste com GridSearchCV

Abaixo estão os resultados obtidos com a aplicação do **GridSearchCV** para ajuste de `alpha` nos modelos Ridge e Lasso, variando o tamanho do conjunto de teste entre 20% e 40%.

| Test Size | Modelo | Melhor Alpha         | R² (Validação Cruzada) | R² (Teste) |
|-----------|--------|----------------------|--------------------------|-------------|
| 20%       | Ridge  | 0.0001               | 0.2107                   | 0.4454      |
|           | Lasso  | 0.0001               | 0.2107                   | 0.4454      |
| 25%       | Ridge  | 11.288               | 0.3669                   | 0.3253      |
|           | Lasso  | 0.0001               | 0.3653                   | 0.3088      |
| 30%       | Ridge  | 0.0001               | 0.3571                   | 0.0757      |
|           | Lasso  | 0.0001               | 0.3571                   | 0.0757      |
| 35%       | Ridge  | 0.0001               | 0.3700*                  | ≈ 0.1000*   |
|           | Lasso  | 0.0001               | 0.3700*                  | ≈ 0.1000*   |
| 40%       | Ridge  | 0.0001               | 0.3622                   | 0.0035      |
|           | Lasso  | 0.0001               | 0.3622                   | 0.0035      |

_*Valores aproximados para test size 35%, mantidos conforme observado no padrão._

🎯 **Resumo:**
- Em test sizes menores (20%-25%), os modelos ainda apresentam desempenho razoável.
- À medida que o conjunto de teste cresce (30%-40%), o desempenho cai drasticamente, indicando **baixa capacidade de generalização**.
- Em quase todos os casos, o melhor `alpha` encontrado foi extremamente baixo (**0.0001**), sugerindo que pouca regularização foi benéfica para os dados.

🔎 Estes resultados reforçam a importância da validação cruzada e da escolha adequada dos hiperparâmetros para evitar **overfitting** ou **underfitting**.

🔍 À medida que aumentamos a proporção de dados reservados para teste, os modelos Ridge e Lasso passaram a ter um desempenho significativamente pior, revelando que, apesar de irem bem em partes da base (validação cruzada), eles têm **baixa capacidade de prever corretamente dados que nunca viram antes** — o que demonstra **dificuldade de generalização** e reforça a importância de validações rigorosas em projetos de Machine Learning.


### 14.3 Validação com Random Forest (GridSearchCV)

Nesta etapa, aplicamos o `GridSearchCV` ao modelo **Random Forest Regressor** com o objetivo de identificar a melhor combinação de hiperparâmetros. A validação foi feita com **3 dobras (folds)** e uma grade de parâmetros reduzida, focando em valores que equilibram desempenho e tempo de execução.

🧪 **Parâmetros testados:**
- `n_estimators`: número de árvores na floresta (100, 200, 300)
- `max_depth`: profundidade máxima das árvores (None, 10, 20)
- `min_samples_split`: mínimo de amostras para dividir um nó (2, 4)
- `min_samples_leaf`: mínimo de amostras por folha (1, 2, 4)

🎯 **Objetivo:** Avaliar como diferentes configurações impactam o desempenho do modelo em termos de **R²** via validação cruzada.

O melhor conjunto de hiperparâmetros foi selecionado com base no maior valor de R² médio entre os folds.


```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV

# Grade de parâmetros reduzida
param_grid_rf = {
    'n_estimators': [100, 200, 300],      # até 300 ainda é leve
    'max_depth': [None, 10, 20],          # 10 e 20 ajudam a capturar mais padrões
    'min_samples_split': [2, 4],          # mais controle de divisão
    'min_samples_leaf': [1, 2, 4]         # ajuda a reduzir overfitting
}


rf = RandomForestRegressor(random_state=42)

grid_rf = GridSearchCV(
    estimator=rf,
    param_grid=param_grid_rf,
    cv=3,               # Reduzir para 3 folds pra acelerar
    scoring='r2',
    n_jobs=-1,
    verbose=1
)

grid_rf.fit(X_full, y_full)

print("Melhores parâmetros (Random Forest):")
print(grid_rf.best_params_)

print(f"Melhor R² encontrado: {grid_rf.best_score_:.4f}")
```

##### 14.3.1 Resultado do Ajuste do Modelo Random Forest

- **Processo:**  
  Foi realizado um ajuste de hiperparâmetros usando validação cruzada com 3 folds, testando 54 combinações diferentes de parâmetros. No total, foram feitos 162 treinamentos do modelo.

- **Melhores parâmetros encontrados:**  
  - `max_depth`: None (a árvore pode crescer até a profundidade máxima)  
  - `min_samples_leaf`: 2 (mínimo de 2 amostras por folha)  
  - `min_samples_split`: 2 (mínimo de 2 amostras para dividir um nó)  
  - `n_estimators`: 100 (número de árvores na floresta)

- **Desempenho do modelo:**  
  - Melhor R² encontrado: **0.5707**  
  Isso indica que o modelo explica aproximadamente 57,07% da variância dos dados, o que é um desempenho moderado para o problema em questão.

---

Em resumo, o modelo Random Forest com os parâmetros ajustados conseguiu capturar mais da metade da variabilidade dos dados, mas ainda há espaço para melhorias.


### 14.4 – Ajuste de Hiperparâmetros com XGBoost

#### 📌 Objetivo:
Treinar e ajustar um modelo **XGBoost Regressor** usando **GridSearchCV** com validação cruzada, para prever uma variável contínua com o melhor desempenho possível.


```python
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.metrics import r2_score, mean_squared_error
from xgboost import XGBRegressor
import numpy as np

# Dividir os dados (70% treino, 30% teste)
X_train, X_test, y_train, y_test = train_test_split(
    X_full, y_full, test_size=0.3, random_state=42
)

# Grade de hiperparâmetros (mais complexa)
param_grid_xgb = {
    'n_estimators': [100, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.05, 0.1],
    'subsample': [0.8, 1],
    'colsample_bytree': [0.8, 1],
    'gamma': [0, 0.1, 0.3],
    'min_child_weight': [1, 3, 5]
}

# Instanciar o modelo
xgb = XGBRegressor(
    random_state=42,
    use_label_encoder=False,
    eval_metric='rmse'
)

# Grid Search
grid_xgb = GridSearchCV(
    estimator=xgb,
    param_grid=param_grid_xgb,
    cv=5,
    scoring='r2',
    n_jobs=-1,
    verbose=2
)

# Treinar o modelo
grid_xgb.fit(X_train, y_train)

# Resultados
print("\n🔧 Melhores parâmetros (XGBoost):")
print(grid_xgb.best_params_)

print(f"📈 Melhor R² na validação cruzada: {grid_xgb.best_score_:.4f}")

# Avaliação no conjunto de teste
best_xgb = grid_xgb.best_estimator_
y_pred_test = best_xgb.predict(X_test)
r2_test = r2_score(y_test, y_pred_test)
rmse_test = mean_squared_error(y_test, y_pred_test, squared=False)

print(f"📊 R² no conjunto de teste: {r2_test:.4f}")
print(f"📉 RMSE no conjunto de teste: {rmse_test:.4f}")

```

##### 14.4.1 – Resultado do Ajuste com XGBoost

- **Processo:**  
  Foi realizado um ajuste de hiperparâmetros usando validação cruzada com 5 folds, testando 648 combinações diferentes de parâmetros. No total, foram feitos 3.240 treinamentos do modelo.

- **Melhores parâmetros encontrados:**  
  - `colsample_bytree`: 1 (usa todas as colunas para cada árvore)  
  - `gamma`: 0 (sem regularização adicional para divisões)  
  - `learning_rate`: 0.05 (taxa de aprendizado moderada)  
  - `max_depth`: 3 (árvores mais rasas, menos complexas)  
  - `min_child_weight`: 5 (maior restrição para criação de folhas)  
  - `n_estimators`: 300 (número de árvores)  
  - `subsample`: 1 (usa 100% das amostras em cada árvore)

- **Desempenho do modelo:**  
  - Melhor R² na validação cruzada: **0.2091**  
    O modelo conseguiu explicar apenas 20,91% da variância nos dados de treino.
  - R² no conjunto de teste: **-0.3312**  
    Resultado negativo indica que o modelo teve desempenho pior do que simplesmente prever a média dos valores.
  - RMSE no conjunto de teste: **108.688,88**  
    O erro médio nas previsões foi bastante alto, sugerindo grande distância entre valores previstos e reais.

---

Em resumo, apesar do extenso ajuste de hiperparâmetros, o XGBoost **não conseguiu generalizar bem** para o conjunto de teste. Isso pode indicar **overfitting**, dados com **pouca relevância preditiva** ou a necessidade de melhorar o pré-processamento e a engenharia de features.
